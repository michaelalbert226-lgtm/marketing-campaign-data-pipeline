```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent
DATASET = BASE_DIR / "datasets" / "marketing_campaign.csv"

df = pd.read_csv(DATASET, sep="\t")
```

```python
df.isna().sum()
```

```python
df[df["Income"].isna()]
```

```python
clean_df = df.dropna(subset=["Income"])
clean_df.shape
```

```python
mean = df["Income"].mean()
filled_df = df.copy()
filled_df["Income"] = filled_df["Income"].fillna(mean)
filled_df.isna().sum()
```

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], format="%d-%m-%Y")
df["Dt_Customer"].dtypes
df["Dt_Customer"].dt.month.head()
df["Dt_Customer"].dt.year.head()
df["Dt_Customer"].dt.day_name().value_counts()
df["Customer_Year"] = df["Dt_Customer"].dt.year
df["Customer_Month"] = df["Dt_Customer"].dt.month
df["Customer_Day"] = df["Dt_Customer"].dt.day

df[[
    "Dt_Customer",
    "Customer_Year",
    "Customer_Month",
    "Customer_Day"
]].head()

df[df["Dt_Customer"].dt.year == 2014].shape[0]

df["Income"] = pd.to_numeric(df["Income"])
df["Income"].dtype

clean_data = pd.to_numeric(
    df["Income"],
    errors="coerce"
)
df["Income"] = clean_data

#finding suspicious birth years
df["Year_Birth"].value_counts().sort_index().head(10)
df[df["Year_Birth"] < 1900]

df[df["Income"] < 0]

```
